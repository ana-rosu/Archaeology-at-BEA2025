```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report

train = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_train.csv')
val   = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_val.csv')

# train = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_train.csv')
# val   = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_val.csv')

# train = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_train.csv')
# val   = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_val.csv')


# train = pd.read_csv('/kaggle/input/actionability-dataset/actionability_train.csv')
# val   = pd.read_csv('/kaggle/input/actionability-dataset/actionability_val.csv')

train_texts = train['response'].fillna('').tolist()
val_texts   = val['response'].fillna('').tolist()
# train_texts = train['last_student_turn'] + train['response'].fillna('').tolist()
# val_texts   = val['last_student_turn'] + val['response'].fillna('').tolist()
y_train     = train['mistake_identification'].values
y_val       = val['mistake_identification'].values

tfidf = TfidfVectorizer(
    analyzer='char',
    ngram_range=(1, 2),
)
X_train = tfidf.fit_transform(train_texts)
X_val   = tfidf.transform(val_texts)

clf = XGBClassifier(
    learning_rate=0.1,
    n_estimators=200,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    use_label_encoder=False,
    eval_metric='mlogloss'
)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_val)

acc_exact = accuracy_score(y_val, y_pred)
f1_exact  = f1_score(y_val, y_pred, average='macro')
print(f"\nExact (3-class) → Acc: {acc_exact:.4f}, Macro F1: {f1_exact:.4f}\n")
print("Classification report (3-class):")
print(classification_report(y_val, y_pred, digits=4))

y_val_bin  = (y_val > 0).astype(int)
y_pred_bin = (y_pred > 0).astype(int)
acc_lenient = accuracy_score(y_val_bin, y_pred_bin)
f1_lenient  = f1_score(y_val_bin, y_pred_bin, average='macro')
print(f"Lenient (2-class) → Acc: {acc_lenient:.4f}, Macro F1: {f1_lenient:.4f}\n")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report
import numpy as np

train = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_train.csv')
val   = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_val.csv')

# train = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_train.csv')
# val   = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_val.csv')

# train = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_train.csv')
# val   = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_val.csv')

# train = pd.read_csv('/kaggle/input/actionability-dataset/actionability_train.csv')
# val   = pd.read_csv('/kaggle/input/actionability-dataset/actionability_val.csv')


train_texts = train['response'].fillna('').tolist()
val_texts   = val['response'].fillna('').tolist()
# train_texts = train['last_student_turn'].fillna('') + ' ' + train['response'].fillna('')
# val_texts   = val['last_student_turn'].fillna('') + ' ' + val['response'].fillna('')
y_train     = train['mistake_identification'].values
y_val       = val['mistake_identification'].values

tfidf = TfidfVectorizer(
    analyzer='char',
    ngram_range=(1, 2),
)
X_train = tfidf.fit_transform(train_texts)
X_val   = tfidf.transform(val_texts)

xgb_clf = XGBClassifier(
    learning_rate=0.1,
    n_estimators=200,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    use_label_encoder=False,
    eval_metric='mlogloss'
)
xgb_clf.fit(X_train, y_train)

lgbm_clf = LGBMClassifier(
    learning_rate=0.1,
    n_estimators=200,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
)
lgbm_clf.fit(X_train, y_train)

xgb_probs = xgb_clf.predict_proba(X_val)
lgbm_probs = lgbm_clf.predict_proba(X_val)

avg_probs = (xgb_probs + lgbm_probs) / 2
y_pred = np.argmax(avg_probs, axis=1)

acc_exact = accuracy_score(y_val, y_pred)
f1_exact  = f1_score(y_val, y_pred, average='macro')
print(f"\nExact (3-class) → Acc: {acc_exact:.4f}, Macro F1: {f1_exact:.4f}\n")
print("Classification report (3-class):")
print(classification_report(y_val, y_pred, digits=4))

=y_val_bin  = (y_val > 0).astype(int)
y_pred_bin = (y_pred > 0).astype(int)
acc_lenient = accuracy_score(y_val_bin, y_pred_bin)
f1_lenient  = f1_score(y_val_bin, y_pred_bin, average='macro')
print(f"Lenient (2-class) → Acc: {acc_lenient:.4f}, Macro F1: {f1_lenient:.4f}\n")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, classification_report
import numpy as np


train = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_train.csv')
val   = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_val.csv')

# train = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_train.csv')
# val   = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_val.csv')

# train = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_train.csv')
# val   = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_val.csv')

# train = pd.read_csv('/kaggle/input/actionability-dataset/actionability_train.csv')
# val   = pd.read_csv('/kaggle/input/actionability-dataset/actionability_val.csv')

# Combine fields into text
train_texts = train['response'].fillna('').tolist()
val_texts   = val['response'].fillna('').tolist()
# train_texts = train['last_student_turn'].fillna('') + ' ' + train['response'].fillna('')
# val_texts   = val['last_student_turn'].fillna('') + ' ' + val['response'].fillna('')
y_train     = train['mistake_identification'].values
y_val       = val['mistake_identification'].values

tfidf = TfidfVectorizer(
    analyzer='word',
    ngram_range=(1, 2),
)
X_train = tfidf.fit_transform(train_texts)
X_val   = tfidf.transform(val_texts)

clf = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_val)

acc_exact = accuracy_score(y_val, y_pred)
f1_exact  = f1_score(y_val, y_pred, average='macro')
print(f"\nExact (3-class) → Acc: {acc_exact:.4f}, Macro F1: {f1_exact:.4f}\n")
print("Classification report (3-class):")
print(classification_report(y_val, y_pred, digits=4))

y_val_bin  = (y_val > 0).astype(int)
y_pred_bin = (y_pred > 0).astype(int)
acc_lenient = accuracy_score(y_val_bin, y_pred_bin)
f1_lenient  = f1_score(y_val_bin, y_pred_bin, average='macro')
print(f"Lenient (2-class) → Acc: {acc_lenient:.4f}, Macro F1: {f1_lenient:.4f}\n")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import HashingVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score, classification_report

train = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_train.csv')
val   = pd.read_csv('/kaggle/input/mistake-identification/mistake_identification_val.csv')

# train = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_train.csv')
# val   = pd.read_csv('/kaggle/input/mistake-loc/mistake_location_val.csv')

# train = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_train.csv')
# val   = pd.read_csv('/kaggle/input/providing-guidance/providing_guidance_val.csv')

# train = pd.read_csv('/kaggle/input/actionability-dataset/actionability_train.csv')
# val   = pd.read_csv('/kaggle/input/actionability-dataset/actionability_val.csv')


train_texts = train['response'].fillna('').tolist()
val_texts = val['response'].fillna('').tolist()
# train_texts = train['last_student_turn'].fillna('') + ' ' + train['response'].fillna('')
# val_texts   = val['last_student_turn'].fillna('') + ' ' + val['response'].fillna('')

y_train = train['mistake_identification'].values
y_val = val['mistake_identification'].values

def spectrum_kernel_vec(train, test=None, start=3, end=5):
    params = {
    'analyzer':'char',
    'lowercase':True,
    'ngram_range':(start, end),
    'binary': True,
    'norm': None,
    'n_features':2**20
    }
    vectorizer = HashingVectorizer(**params)
    X_train = vectorizer.fit_transform(train)
    if test is not None:
        X_test = vectorizer.transform(test)
        return cosine_similarity(X_test, X_train)
    else:
        return cosine_similarity(X_train, X_train)


K_train_train = spectrum_kernel_vec(train_texts)
K_val_train = spectrum_kernel_vec(train_texts, val_texts)

svc = SVC(kernel='precomputed')
svc.fit(K_train_train, y_train)

y_pred = svc.predict(K_val_train)

acc_exact = accuracy_score(y_val, y_pred)
f1_exact = f1_score(y_val, y_pred, average='macro')

y_val_binary = (y_val > 0).astype(int)
y_pred_binary = (y_pred > 0).astype(int)

acc_lenient = accuracy_score(y_val_binary, y_pred_binary)

def lenient_f1(y_true, y_pred):
    return f1_score((y_true > 0).astype(int), (y_pred > 0).astype(int), average='macro')

f1_lenient = lenient_f1(y_val, y_pred)

# Print results
print(f"\nExact (3-class) → Acc: {acc_exact:.4f}, Macro F1: {f1_exact:.4f}")
print(f"Lenient (2-class) → Acc: {acc_lenient:.4f}, Macro F1: {f1_lenient:.4f}\n")
```
